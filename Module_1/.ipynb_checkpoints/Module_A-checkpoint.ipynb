# Дополнение набора данных дополнительными столбцами

## Добовляем к каждому файлу в таблице длительность, частоту дискретизации, число каналов и среднюю энтропию Винера

```python
folder = Path("../dataset/Soundwel Dataset - Audio and Spectrograms/")
```

```python
colum = []

for file in folder.iterdir():
    if file.suffix == ".wav":
        audio, sr = librosa.load(file.absolute(), sr=None, mono=False)
        sample_rate = sr # Частота дискретизации
        channels = 1 if audio.ndim == 1 else audio.shape[0] # Количество каналов
        duration = audio.shape[-1] / sr # Длительность аудио

        mono = audio.mean(axis=0) if channels > 1 else audio
        # Спектральная плоскостность
        flatness = librosa.feature.spectral_flatness(
            y=mono,
            n_fft=2048,
            hop_length=512,
            power=2.0
        )[0]
        
        # Средняя энтропия Винера
        mean_wiener = np.mean(flatness)
        colum.append({"sample_rate": sample_rate, "channels": channels, "duration": duration, "mean_wiener": mean_wiener})
```

```python
colum
```

## Добавляем полученные данные к таблице

```python
df = pd.concat([df, pd.DataFrame(colum)], axis=1)
```

```python
df
```

## Производим аугментацию аудио-файлов в датасете

```python
class AudioDataAugmentation:
    def __init__(self, audio: np.array):
        if audio.ndim == 2:
            audio = audio.mean(axis=1)
        self.audio = audio
        self.rng = np.random.default_rng(42)
    def random_gain(self, low_db=-4, high_db=4) -> np.array:
        # 1. Случайно выбираем изменение в dB
        gain_db = self.rng.uniform(low_db, high_db)
    
        # 2. Переводим децибелы в множитель
        factor = 10 ** (gain_db / 20)
    
        # 3. Изменяем амплитуду
        y = self.audio * factor
    
        # 4. Защищаем сигнал от выхода за пределы
        peak = np.max(np.abs(y)) if y.size else 0.0
    
        if peak > 0.99:
            y = y * (0.99 / peak)
    
        return y.astype(np.float32)
    def gaussian_noise_snr(self, snr_db):
        # 1. Измеряем RMS полезного сигнала
        signal_rms = np.sqrt(
            np.mean(self.audio.astype(np.float64) ** 2)
        )
    
        # Не усиливаем шум на полностью тихих фрагментах
        if signal_rms < 1e-8:
            return self.audio.copy()
    
        # 2. Генерируем стандартный нормальный шум
        noise = self.rng.normal(size=self.audio.shape)
    
        # 3. Нормируем RMS шума до 1
        noise_rms = np.sqrt(np.mean(noise ** 2))
        noise = noise / (noise_rms + 1e-12)
    
        # 4. Вычисляем нужную громкость шума
        target_noise_rms = (
            signal_rms / (10 ** (snr_db / 20))
        )
    
        noise = noise * target_noise_rms
    
        # 5. Складываем сигналы
        y = self.audio + noise
    
        return y.astype(np.float32)
```

```python
folder = Path("../dataset/Soundwel Dataset - Audio and Spectrograms/")
table = []
for file in folder.iterdir():
    if file.suffix == ".wav":
        audio, sr = sf.read(file, dtype="float32")
        ADA = AudioDataAugmentation(audio)
        aug1 = ADA.gaussian_noise_snr(20)
        aug2 = ADA.random_gain()
        sf.write(f"../dataset/Augmentation/{file.stem}_1.wav", aug1, sr) # Аугментация по громкости
        sf.write(f"../dataset/Augmentation/{file.stem}_2.wav", aug2, sr) # Аугментация по шуму
        table.append({"Audio Filename": f"{file.stem}_1.wav", "Age Category": int(df[df["Audio Filename"] == file.name]["Age Category"].iloc[0]),
                      "Valence": int(df[df["Audio Filename"] == file.name]["Valence"].iloc[0]),
                      "duration": float(df[df["Audio Filename"] == file.name]["duration"].iloc[0]),
                      "mean_wiener": float(df[df["Audio Filename"] == file.name]["mean_wiener"].iloc[0])})
        table.append({"Audio Filename": f"{file.stem}_2.wav", "Age Category": int(df[df["Audio Filename"] == file.name]["Age Category"].iloc[0]),
                      "Valence": int(df[df["Audio Filename"] == file.name]["Valence"].iloc[0]),
                      "duration": float(df[df["Audio Filename"] == file.name]["duration"].iloc[0]),
                      "mean_wiener": float(df[df["Audio Filename"] == file.name]["mean_wiener"].iloc[0])})
```

```python
pd.DataFrame(table).to_csv("../dataset/AugDatasetKey.csv")
```

```python
with sqlite3.connect("datasets.sqlite") as conn:
    df.to_sql(name="main_dataset", con=conn, if_exists="replace", index=False)
    pd.DataFrame(table).to_sql(name="augmentation_dataset", con=conn, if_exists="replace", index=False)
    pd.read_csv("../dataset/SoundwelDatasetKey.csv").to_sql(name="initial_table", con=conn, if_exists="replace", index=False)
```
